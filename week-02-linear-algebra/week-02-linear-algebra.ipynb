```python
import numpy as np
import matplotlib.pyplot as plt
```

## Matrix Inverse

The inverse of a matrix A is another matrix A⁻¹ such that:

A × A⁻¹ = I

Where I is the identity matrix.

In Machine Learning:
- Used in normal equation for Linear Regression
- Helps solve systems of linear equations


```python
A = np.array([[2, 1],
              [1, 2]])

A_inv = np.linalg.inv(A)

print("Matrix A:\n", A)
print("\nInverse of A:\n", A_inv)

```

## Rank of a Matrix

The rank of a matrix is the number of linearly independent rows (or columns).

Why it matters in ML:
- Determines if a system of equations has a unique solution
- Helps detect multicollinearity
- Important in regression problems


```python
rank = np.linalg.matrix_rank(A)
print("Rank of A:", rank)
```

## Eigenvalues and Eigenvectors

For a matrix A:

A * v = λ * v

Where:
- v = eigenvector
- λ = eigenvalue

Eigenvectors represent directions.
Eigenvalues represent scaling along those directions.

Used in:
- Principal Component Analysis (PCA)
- Dimensionality Reduction
- Spectral Clustering


```python
eigenvalues, eigenvectors = np.linalg.eig(A)

print("Eigenvalues:\n", eigenvalues)
print("Eigenvectors:\n", eigenvectors)
```

## Linear Transformation (Visualization)

A matrix transforms vectors in space.

This is how neural networks and many ML models operate:
- Input data → matrix multiplication → transformed output


```python
vector = np.array([1, 0])
transformed = A @ vector

plt.quiver(0, 0, vector[0], vector[1], angles='xy', scale_units='xy', scale=1, label="Original")
plt.quiver(0, 0, transformed[0], transformed[1], angles='xy', scale_units='xy', scale=1, label="Transformed")

plt.xlim(-3, 3)
plt.ylim(-3, 3)
plt.grid()
plt.title("Linear Transformation Visualization")
plt.show()
```

## Why Linear Algebra is Critical in Machine Learning

- Datasets are represented as matrices
- Model parameters are vectors
- Training involves matrix multiplication
- PCA uses eigenvalues and eigenvectors
- Neural networks rely heavily on linear transformations

Without linear algebra, machine learning cannot exist.
