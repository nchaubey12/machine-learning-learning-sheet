# Week 01 - Python & NumPy for Machine Learning

This notebook covers:
- NumPy arrays
- Matrix operations
- Vectorization
- Basic statistics


```python
import numpy as np
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

print("Array A:", a)
print("Array B:", b)
```

```python
dot_product = np.dot(a, b)
print("Dot Product:", dot_product)
```

```python
A = np.array([[1, 2], [3, 4]])
B = np.array([[5, 6], [7, 8]])

result = np.matmul(A, B)
print("Matrix Multiplication:\n", result)
```

```python
data = np.array([10, 20, 30, 40, 50])

print("Mean:", np.mean(data))
print("Variance:", np.var(data))
```
